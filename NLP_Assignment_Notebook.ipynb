### Task 4: Simple Movie Review Sentiment Checker (Basic Python)


```python
# Task 4: Simple Sentiment Checker
def check_sentiment(review):
    review_lower = review.lower()
    has_good = 'good' in review_lower
    has_bad = 'bad' in review_lower
    
    if has_good and not has_bad:
        return 'Positive'
    elif has_bad and not has_good:
        return 'Negative'
    else:
        return 'Neutral'

# Test with reviews
sample_reviews = [
    "The cinematography was really good and acting was top notch!",
    "The screenplay was so bad that I left the theater midway.",
    "The movie was an average watch, neither exciting nor terrible."
]

for rev in sample_reviews:
    print(f"Review: '{rev}' -> Sentiment: {check_sentiment(rev)}")
```

### Task 4: Remove Stopwords using NLTK


```python
# Task 4: Stopwords Removal
import nltk
from nltk.corpus import stopwords

try:
    stop_words = set(stopwords.words('english'))
except LookupError:
    nltk.download('stopwords')
    stop_words = set(stopwords.words('english'))

def remove_stopwords(message):
    words = message.split()
    return " ".join([w for w in words if w.lower() not in stop_words])

whatsapp_msg = "Hey buddy, are we still meeting at the cafe for the project discussion this evening?"
print("Original:", whatsapp_msg)
print("Filtered:", remove_stopwords(whatsapp_msg))
```

### Task 3: Word Frequency Count across WhatsApp Messages


```python
# Task 3: Word Frequency Dictionary
import string

messages = [
    "Hey, are you coming to the college library today?",
    "Yes, I will be at the library by 2 PM.",
    "Great, let's finish the NLP project assignment together at the library.",
    "Don't forget to bring the project notes!"
]

stopwords_set = {"the", "is", "at", "to", "you", "i", "be", "by", "are", "and", "a", "in", "it", "of", "for"}
freq_map = {}

for msg in messages:
    clean = msg.translate(str.maketrans('', '', string.punctuation)).lower()
    for w in clean.split():
        if w not in stopwords_set:
            freq_map[w] = freq_map.get(w, 0) + 1

for word, count in sorted(freq_map.items(), key=lambda x: x[1], reverse=True):
    print(f"{word:<15}: {count}")
```

### Task 4: TF-IDF Vectors & 'quality' Score on Flipkart Reviews


```python
# Task 4: TF-IDF Inspection
from sklearn.feature_extraction.text import TfidfVectorizer

flipkart_reviews = [
    "The sound quality is crisp and battery backup is top notch.",
    "Terrible product, poor build quality and arrived completely damaged.",
    "Super fast shipping, highly satisfied with this purchase.",
    "Premium quality material and premium finish, worth the price."
]

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(flipkart_reviews)
feature_names = list(tfidf.get_feature_names_out())

quality_idx = feature_names.index("quality")
for i, rev in enumerate(flipkart_reviews, 1):
    val = tfidf_matrix[i - 1, quality_idx]
    print(f"Review {i}: '{rev}'\n-> TF-IDF score for 'quality': {val:.4f}\n")
```

### Task 5: BoW vs. TF-IDF Differences & Real-World Use Case
1. **Raw Counts vs. Document Informativeness:** BoW simply tallies frequencies, meaning non-informative frequent terms dominate. TF-IDF penalizes words that appear across all documents and rewards terms specific to single documents.
2. **Discrimination Ability:** BoW treats every occurrence identically; TF-IDF gives distinct semantic weight to distinguishing words.

*Real-World Example:* In Flipkart/Amazon product searches (e.g., *"noise cancelling over-ear headphones"*), generic words like *"headphones"* appear in every review and dominate BoW. TF-IDF elevates *"noise-cancelling"* and *"over-ear"*, ranking the most relevant products to the top.

### Task 2: Limitation of Traditional NLP on WhatsApp Chat
Traditional models like Bag of Words (BoW) and TF-IDF completely ignore word order, sentence context, and syntactic structure. In casual WhatsApp conversations where sarcasm, slang, negations, or emojis abound (e.g., *"Oh great, my flight got cancelled!"*), BoW simply counts the positive word *"great"* and misclassifies the message as positive. Furthermore, traditional models fail on chat abbreviations (e.g., *"idk"*, *"tbh"*, *"wfh"*) because they cannot model contextual semantics or subtle relational meanings across words.

### Task 5: AI-Generated Attention Mechanism & Explanation
**Prompt to ChatGPT / Copilot:**
> *"Show a basic attention calculation example in Python demonstrating the scaled dot-product attention mechanism with three short sentences as input, and print the attention score matrix."*


```python
# Task 5: Attention Mechanism Demonstration
import numpy as np

def softmax(x):
    e_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return e_x / np.sum(e_x, axis=-1, keepdims=True)

def compute_attention(embeddings):
    d_k = embeddings.shape[1]
    scores = np.matmul(embeddings, embeddings.T) / np.sqrt(d_k)
    return softmax(scores)

sentences = [
    "1. AI powers smart recommendations.",
    "2. Machine learning creates intelligent algorithms.",
    "3. Today is a sunny and warm morning."
]

# Simulated semantic embeddings (Sentences 1 & 2 share AI topic, Sentence 3 is weather)
sentence_embeddings = np.array([
    [1.2, 0.9, 0.1, 0.05],
    [1.1, 0.85, 0.15, 0.02],
    [0.05, 0.1, 1.3, 0.9]
])

weights = compute_attention(sentence_embeddings)
print("Attention Score Matrix:")
print(np.round(weights, 4))
```

**Explanation of Attention Scores:**  
The attention scores represent normalized probability weights (each row sums to 1.0) that quantify how much focus or 'relevance' one sentence allocates to another when interpreting meaning. Sentences 1 and 2 assign very high attention weights to each other (~40-43% each) because they share strong semantic concepts (AI/algorithms), while allocating minimal attention to Sentence 3 (~16-17%), which pertains to unrelated weather.
